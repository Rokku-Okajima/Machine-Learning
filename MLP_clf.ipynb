```python
from sklearn.datasets import fetch_openml

# Download Fashion MNIST Dataset
fashion_MNIST = fetch_openml(name="Fashion-MNIST", as_frame=True)
targets = fashion_MNIST.target.astype(int)
```

```python
# split into train and test sets 
X_train, y_train =  fashion_MNIST.data[:60000], targets[:60000]
X_test, y_test = fashion_MNIST.data[60000:], targets[60000:]
```

```python
class_names = ["T-shirt/top",  "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle Boot"]
```

```python
class_names[y_train[0]]
```

```python
# Build the Classification MLP 
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler

mlp_clf = MLPClassifier(hidden_layer_sizes=[500, 100], verbose=True, early_stopping=True, # two hidden layer with 500 and 100
                        random_state=42, activation="relu", learning_rate="adaptive")
```

```python
from sklearn.pipeline import make_pipeline
pipline = make_pipeline(MinMaxScaler(), mlp_clf)
pipline.fit(X_train, y_train)
accuarcy = pipline.score(X_test, y_test)
```

```python
model_accuracy = mlp_clf.best_validation_score_
print(f"The MLP Classification accuracy: {model_accuracy:.2f}%")
```
